```python
import pandas as pd
```

```python
df = pd.concat([pd.read_csv("raw/recursos_aa.csv"),
                pd.read_csv("raw/recursos_pp.csv")]).reset_index(drop=True)
```

```python
df = df.rename(columns={'dimensión': 'dimension',
                        'tema': 'theme',
                        'tipo': 'type',
                        'detalle': 'detail',
                        'código_recurso': 'standard_code'})
```

```python
df["standard_code"] = df["standard_code"].str.findall(r"[A-Za-z0-9]+")
```

```python
df = df.explode('standard_code')
```

```python
df = df.reset_index(drop=True)
```

```python
df.sort_values(by=['standard_code'])
```

```python
# Combine url columns into a single 'urls' column as a dictionary
url_cols = ["url_web", "url_curso", "url_pdf", "url_word", "url_excel"]
df["urls"] = df[url_cols].apply(lambda row: {k.split('_')[1]: v for k, v in row.items() if pd.notnull(v) and v != ''}, axis=1)
df = df.drop(columns=url_cols)
```

```python
# Create a 'resource_code' column as a unique identifier for each resource
# Example: use the index or a combination of columns (customize as needed)
df["resource_code"] = df.index.astype(str).str.zfill(6)
df.head()
```

```python
df.to_csv("clean/resources.csv", index=False)
```
