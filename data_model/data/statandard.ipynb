```python
import pandas as pd
```

```python
input_file = "raw/registros_pp.csv"
output_file = "clean/standard_pp.csv"
```

```python
df_resources = pd.read_csv("clean/resources.csv")
```

```python
df_linked_resources = df_resources.groupby('standard_code')['resource_code'].apply(list).reset_index()
df_linked_resources = df_linked_resources.rename(columns={'resource_code': 'linked_resources'})
```

```python
valid_answers = ["Si cumplo", "No cumplo", "No cumplo, pero me es factible", "No aplica a mi sistema productivo"]
```

```python
mapping = {"Código":"standard_code",
           "Nivel": "level",
           "Puntos": "points",
           "Dimensión": "dimension",
           "Tema": "theme",
           "Buena Práctica": "good_practice",
           "Acción": "action",
           "Medio de Verificación": "verification_detail",
           "IA": "verification_medium",
           "Link": "link",
           "Link Recursos": "link"}
```

```python
df = pd.read_csv(input_file).drop(columns=["N°"]).rename(columns=mapping)
```

```python
df['valid_answers'] = [valid_answers.copy() for _ in range(len(df))]
```

```python
df = df.merge(df_linked_resources)
```

```python
df.to_csv(output_file, index=False, encoding="utf-8-sig")
```
